### Olympic Games Data Analysis

### Importing Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Loading Datasets

```python
# to import upload .csv file using pandas?
athletes = pd.read_csv('olympics.csv')
athletes
```

```python
# shows the first five records of the dataset
athletes.head()
```

```python
# to display last five rows
athletes.tail()
```

### Basic Functions

### check Dataset Dimensions

```python
# to return no.of rows and columns
athletes.shape
```

### Get column names

```python
athletes.columns
```

### Remove Unnecessary column

```python
athletes.drop("Unnamed: 0",axis=1,inplace=True)
athletes
```

### Dataset Information

```python
athletes.info()
```

### Statistical Summary

```python
# gives mean,standard deviation,minimum and maximum values
athletes.describe()
```

### Data Cleaning

### Missing values

```python
athletes.isnull().sum()
```

### check duplicate values

```python
athletes.duplicated().sum()
```

```python
athletes.drop_duplicates(inplace=True)
athletes
```

### Data Filtering

### Medal winners

```python
medal_winners=athletes[athletes["Medal"].notna()]
medal_winners
```

### Female Athletes

```python
female=athletes[athletes["Sex"]=="F"]
female
```

### Summer Olympics

```python
summer=athletes[athletes["Season"]=="Summer"]
summer
```

```python
# Athletes older than 30

age = athletes[athletes["Age"]>30]
age
```

```python
# Top 10 medal winning countries

medals = athletes.dropna(subset=["Medal"])
medals["Team"].value_counts().head(10)
```

### Data Sorting

```python
# sorting by age
athletes.sort_values(by="Age")
```

```python
# sorting by year
athletes.sort_values(by="Year",ascending=False)
```

```python
# sorting by names of athletes
athletes.sort_values(by="Name")
```

### Data Grouping

```python
# Average age by gender
athletes.groupby("Sex")["Age"].mean()
```

```python
# no.of athletes by country
athletes.groupby("Team")["Name"].count()
```

```python
# no.of athletes by sports
athletes.groupby("Sport")["Name"].count()
```

```python
# no.of athletes by year
athletes.groupby("Year")["Name"].count()
```

### Aggregate Functions

```python
# average age
athletes["Age"].mean()
```

```python
# maximum age
athletes["Age"].max()
```

```python
# minimum age
athletes["Age"].min()
```

```python
# sum of ages
athletes["Age"].sum()
```

```python
# count
athletes["Age"].count()
```

```python
# median
athletes["Age"].median()
```

```python
# standard deviation
athletes["Age"].std()
```

```python
athletes.groupby("Sex")["Age"].agg(["mean","min","max","count"])
```

### Data Visualization

### Gender Distribution

```python
athletes["Sex"].value_counts()
plt.figure(figsize=(6,4))
sns.countplot(x="Sex", data=athletes)
plt.title("Gender Distribution")
plt.show()
```

### Age Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(athletes["Age"].dropna(), bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Athletes")
plt.show()
```

### Top 10 Countries by Participation

```python
plt.figure(figsize=(10,5))
athletes["Team"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Countries by Participation")
plt.xlabel("Country")
plt.ylabel("Number of Athletes")
plt.show()
```

### Top 10 Sports

```python
plt.figure(figsize=(10,5))
athletes["Sport"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Sports")
plt.xlabel("Sport")
plt.ylabel("Number of Athletes")
plt.show()
```

### Summer vs Winter Olympics

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Season", data=athletes)
plt.title("Summer vs Winter Olympics")
plt.show()
```

### Medal Distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Medal", data=athletes)
plt.title("Medal Distribution")
plt.show()
```

### Top 10 medal winning countries

```python
plt.figure(figsize=(10,5))
medals["Team"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Medal Winning Countries")
plt.xlabel("Country")
plt.ylabel("Number of Medals")
plt.show()
```

### Top 10 Athletes by no.of medals

```python
plt.figure(figsize=(10,5))
medals["Name"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Athletes")
plt.xlabel("Athlete")
plt.ylabel("Medals")
plt.xticks(rotation=90)
plt.show()
```

### Olympic participation over the years

```python
plt.figure(figsize=(12,5))
athletes.groupby("Year")["Name"].count().plot()
plt.title("Olympic Participation Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Athletes")
plt.show()
```
